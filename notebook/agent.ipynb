```python
from agent_framework.foundry import FoundryAgent
from agent_framework import AgentSession, Agent, AgentRunInputs
from azure.identity.aio import AzureCliCredential, OnBehalfOfCredential
from dotenv import load_dotenv
import aiohttp
import os
import json
```

```python
load_dotenv(override=True)

FOUNDRY_PROJECT_ENDPOINT=os.getenv('FOUNDRY_PROJECT_ENDPOINT')
FOUNDRY_AGENT_NAME=os.getenv('FOUNDRY_AGENT_NAME')
FOUNDRY_AGENT_VERSION=os.getenv('FOUNDRY_AGENT_VERSION')
CLIENT_ID = os.getenv('CLIENT_ID')
TENANT_ID = os.getenv('TENANT_ID')
CLIENT_SECRET = os.getenv('CLIENT_SECRET')
TOKEN=os.getenv('TOKEN')
```

```python
agent = FoundryAgent(
    project_endpoint=FOUNDRY_PROJECT_ENDPOINT,
    agent_name=FOUNDRY_AGENT_NAME,
    agent_version=FOUNDRY_AGENT_VERSION,
    allow_preview=True,
    credential=OnBehalfOfCredential(TENANT_ID,
                                    CLIENT_ID,
                                    client_secret=CLIENT_SECRET,
                                    user_assertion=TOKEN)
)    
```

```python
session = agent.create_session()
```

```python
session.session_id
```

```python
agent_chat = FoundryAgent(
    project_endpoint=FOUNDRY_PROJECT_ENDPOINT,
    agent_name=FOUNDRY_AGENT_NAME,
    agent_version=FOUNDRY_AGENT_VERSION,
    allow_preview=True,
    credential=OnBehalfOfCredential(TENANT_ID,
                                    CLIENT_ID,
                                    client_secret=CLIENT_SECRET,
                                    user_assertion=TOKEN)
)  
```

```python
old_session = agent_chat.get_session(service_session_id=session.service_session_id,                                         
                                     session_id=session.session_id)
```

```python
payload = json.dumps({ 'username': 'Hugo Girard', 'prompt': 'Give me the last 10 transactions' })

response = await agent_chat.run(payload,session=old_session)
```
